# Palmer Penguins

```python
import logging
from datetime import UTC, datetime

from palmerpenguins import load_penguins
from plotnine import (
    aes,
    facet_wrap,
    geom_boxplot,
    geom_jitter,
    geom_point,
    geom_smooth,
    ggplot,
    labs,
    scale_color_manual,
    scale_fill_manual,
    theme,
    theme_minimal,
)

logging.basicConfig(
    level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s", force=True
)
logger = logging.getLogger("jupyter-hello")

raw = load_penguins()
penguins = raw.dropna()
logger.info(
    "Loaded %d penguins, dropped %d with missing values",
    len(penguins),
    len(raw) - len(penguins),
)
palette = {"Adelie": "#FF8C00", "Chinstrap": "#A034F0", "Gentoo": "#159090"}

print(f"Rendered at {datetime.now(UTC):%Y-%m-%d %H:%M:%S} UTC")
```

## Species summary

```python
logger.info("Summarizing %d species", penguins["species"].nunique())
(
    penguins.groupby("species")
    .agg(
        count=("species", "size"),
        bill_length_mm=("bill_length_mm", "mean"),
        bill_depth_mm=("bill_depth_mm", "mean"),
        flipper_length_mm=("flipper_length_mm", "mean"),
        body_mass_g=("body_mass_g", "mean"),
    )
    .round(1)
)
```

## Bill length vs. depth

```python
logger.info("Plotting bill length vs. depth")
(
    ggplot(penguins, aes("bill_length_mm", "bill_depth_mm"))
    + geom_smooth(method="lm", color="#444444", linetype="dashed", se=False)
    + geom_point(aes(color="species"), alpha=0.6, size=2)
    + geom_smooth(aes(color="species"), method="lm", se=False, size=1.2)
    + scale_color_manual(values=palette)
    + labs(x="Bill length (mm)", y="Bill depth (mm)", color="Species")
    + theme_minimal()
    + theme(figure_size=(8, 5))
)
```

## Body mass by species and sex

```python
logger.info("Plotting body mass by species and sex")
(
    ggplot(penguins, aes("sex", "body_mass_g", fill="species"))
    + geom_boxplot(alpha=0.4, outlier_shape="", width=0.6)
    + geom_jitter(aes(color="species"), width=0.15, alpha=0.6, size=1.5)
    + facet_wrap("species")
    + scale_fill_manual(values=palette)
    + scale_color_manual(values=palette)
    + labs(x="", y="Body mass (g)")
    + theme_minimal()
    + theme(figure_size=(9, 4), legend_position="none")
)
```
