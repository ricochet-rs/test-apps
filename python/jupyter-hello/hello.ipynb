## A simple plot

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x + np.random.uniform(0, 2 * np.pi))

plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.title("Sine Wave with a Random Phase")
plt.xlabel("x")
plt.ylabel("sin(x + phase)")
plt.grid(True)
plt.show()
```
